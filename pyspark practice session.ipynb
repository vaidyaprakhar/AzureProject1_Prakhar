```python
df_cust = spark.table("samples.tpch.customer")
display(df_cust.limit(10))
```

```python
df_odr = spark.table("samples.tpch.orders")
display(df_odr.limit(10))
```

```python
from pyspark.sql.functions import *

df_join = df_odr.join(
    df_cust,
    df_odr.o_custkey == df_cust.c_custkey,
    "left"
).orderBy(df_odr.o_orderdate.desc())
#display(df_join.limit(10)) 

df_select = df_join.select(
    df_odr.o_orderkey.alias("order_id"),
    df_odr.o_orderdate.alias("order_date"),
    df_odr.o_totalprice.alias("order_price"),
    df_odr.o_orderpriority.alias("order_priority"),
    df_odr.o_shippriority.alias("order_shippriority"),
    df_odr.o_comment.alias("order_comment"),
    df_cust.c_custkey.alias("customer_id"),
    df_cust.c_name.alias("customer_name"),
    df_cust.c_address.alias("customer_address"),
    df_cust.c_phone.alias("customer_phone"),
    df_cust.c_acctbal.alias("customer_balance"),
    df_cust.c_mktsegment.alias("customer_segment"),
    df_cust.c_comment.alias("customer_comment")
)
#display(df_select.limit(10))
df_select.write.mode("overwrite").saveAsTable("pyspark_practice_session.orders_customer")

display(spark.table("pyspark_practice_session.orders_customer").limit(10))

```
